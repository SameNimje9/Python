```python
import matplotlib.pyplot as plt
import numpy as np
import random
```

```python
ml_student_age = np.random.randint(18,45,(100))
py_student_age = np.random.randint(15,40,(100))
```

```python
print(ml_student_age)
print(py_student_age)
```

```python
plt.hist(ml_student_age)

plt.title("ML Students Age Histogarms")
plt.xlabel("Student age category")
plt.ylabel("No. Student age")
plt.show()
```

```python
bins = [15,20,25,30,35,40,45]
plt.figure(figsize = (16,9))
plt.hist([ml_student_age,py_student_age],bins,rwidth = 0.8,histtype ="bar",orientation = "vertical",
        color=['g','m'],label = ["ML Student","Py Student"])
#plt.hist(py_student_age,bins,rwidth = 0.8,histtype ="bar",orientation = "vertical",color='b',label = "Py Student")
plt.title("ML Students age histograms")
plt.xlabel("Student age category")
plt.ylabel("No. Student age")
plt.legend()
plt.show()
```

```python
from matplotlib import style
style.use("ggplot")
plt.figure(figsize = (16,9))
plt.hist([ml_student_age,py_student_age],bins,rwidth = 0.8,histtype ="bar",orientation = "vertical",
        color=['g','m'],label = ["ML Student","Py Student"])
#plt.hist(py_student_age,bins,rwidth = 0.8,histtype ="bar",orientation = "vertical",color='b',label = "Py Student")
plt.title("ML Students age histograms")
plt.xlabel("Student age category")
plt.ylabel("No. Student age")
plt.legend()
plt.show()
```
