# Ploting Bar Chart

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib import style
```

```python
Classes = ["Python","R","AI","ML","DS"]
Class1_student = [30,10,20,25,10]
Class2_student = [40,5,20,20,10]
Class3_student = [35,5,30,15,14]
```

```python
plt.bar(Classes,Class1_student)
```

```python
plt.barh(Classes,Class1_student)
```

```python
plt.bar(Classes,Class1_student,width=0.5,align="edge",color = 'g',
        edgecolor ='y',linewidth =5,alpha = 0.9,linestyle = "--",
        label="Class 1 student")
```

```python
style.use("ggplot")
plt.figure(figsize=(16,9))
plt.bar(Classes,Class1_student,width=0.5,align="edge",color = 'g',
        edgecolor ='m',linewidth =5,alpha = 0.9,linestyle = "--",
        label="Class 1 student")
```

```python
style.use("ggplot")
plt.figure(figsize=(16,9))
plt.barh(Classes,Class1_student,align="edge",color = 'g',
        edgecolor ='m',linewidth =5,alpha = 0.9,linestyle = "--",
        label="Class 1 student")
```

```python
style.use("ggplot")
plt.figure(figsize=(16,9))
plt.bar(Classes,Class1_student,align="edge",color = 'g',
        edgecolor ='m',linewidth =5,alpha = 0.9,linestyle = "--",
        label="Class 1 student")
plt.title("Bar Chart of IAIP Class",fontsize = 18)
plt.xlabel("Classes",fontsize = 15)
plt.ylabel("No. of Student",fontsize = 15)
plt.show()
```

```python
style.use("ggplot")
plt.figure(figsize=(16,9))
plt.bar(Classes,Class1_student,width=0.5,color = 'b',
        label="Class 1 student")

plt.bar(Classes,Class2_student,width=0.5,color = 'g',
        label="Class 2 student")

plt.title("Bar Chart of IAIP Class",fontsize = 18)
plt.xlabel("Classes",fontsize = 15)
plt.ylabel("No. of Student",fontsize = 15)
plt.show()
```

```python
style.use("ggplot")
plt.figure(figsize=(16,9))

Classes_index = np.arange(len(Classes))

width=0.2
plt.bar(Classes_index,Class1_student,width,color = 'b',
        label="Class 1 student")

plt.bar(Classes_index+width,Class2_student,width,color = 'g',
        label="Class 2 student")

plt.bar(Classes_index+width+width,Class3_student,width,color = 'm',
        label="Class 3 student")

plt.xticks(Classes_index+width, Classes,rotation=10)
plt.title("Bar Chart of IAIP Class",fontsize = 18)
plt.xlabel("Classes",fontsize = 15)
plt.ylabel("No. of Student",fontsize = 15)
plt.show()
```

```python
style.use("ggplot")
plt.figure(figsize=(16,9))

Classes_index = np.arange(len(Classes))

width=0.2
plt.barh(Classes_index,Class1_student,width,color = 'b',
        label="Class 1 student")

plt.barh(Classes_index+width,Class2_student,width,color = 'g',
        label="Class 2 student")

plt.barh(Classes_index+width+width,Class3_student,width,color = 'm',
        label="Class 3 student")

plt.yticks(Classes_index+width, Classes,rotation=10)
plt.title("Bar Chart of IAIP Class",fontsize = 18)
plt.ylabel("Classes",fontsize = 15)
plt.xlabel("No. of Student",fontsize = 15)
plt.legend()
plt.show()
```
