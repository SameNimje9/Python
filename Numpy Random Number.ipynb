# Rand() - the function is used to generate a random value between 0 to 1.
# randn() - the function is used to generate a random value close to zero.This may return positive or negative number as well.
# ranf() - The function for doing random sampling in numpy. It return an array of specified shape and filld it with random floats in the half open interval [0.0,1.0).
# randint()- the function is used to generate a random number between a given range

# Random

```python
import numpy as np
```

```python
var = np.random.rand(4)
print(var)
```

```python
var1 =np.random.rand(2,5)
print(var1)
```

# randn()

```python
var2 = np.random.randn(5)
print(var2)
```

# ranf()

```python
var3 = np.random.ranf(4)
print(var3)
```

# randint()

```python
var4 = np.random.randint(5,20,5)
print(var4)
```
