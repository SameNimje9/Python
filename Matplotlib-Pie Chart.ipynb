```python
import matplotlib.pyplot as plt
```

```python
classes=["Python",'R','Machine Learning',"Artificial Intelligence","Data Science"]
Class1_student =[45,15,35,25,30]
plt.pie([1])
plt.show()
```

```python
classes=["Python",'R','Machine Learning',"Artificial Intelligence","Data Science"]
Class1_student =[45,15,35,25,30]
explode = [0,0.03,0.1,0,0]
colors=['c','b','r','y','g']

plt.figure(figsize=(6,2))
wedgeprops ={"linewidth":4,'width':3,"edgecolor":'k'}
textprops={"fontsize":15}
plt.pie(Class1_student,
        labels=classes,
        explode=explode,
        colors=colors, 
        autopct ="%0.2f%%",
        shadow=True,
        radius = 1.4,
        startangle = 270,
        textprops=textprops,
       labeldistance =1.4,
       counterclock=True,
       wedgeprops = wedgeprops,
       rotatelabels = True,
       frame=True)
plt.legend()
plt.show()
```
