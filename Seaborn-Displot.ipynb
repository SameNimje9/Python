```python
sns.displot(tips_df["total_bill"],hue_norm=False,label="Total Bill")
plt.title("Histogram of total bill")
plt.legend()
```

```python
plt.figure(figsize=(16,9))
sns.set()
sns.displot(tips_df["total_bill"],hue_norm=False,label="Total Bill")
plt.title("Histogram of total bill")
plt.legend()
```

```python
tips_df.total_bill.sort_values()
```

```python
bins=[1,5,10,15,25,30,35,40,45,50,55]
plt.figure(figsize=(16,9))
sns.set()
sns.displot(tips_df["total_bill"],bins=bins)
plt.xticks(bins)
plt.title("Histogram of total bill")
plt.show()
```

```python
plt.figure(figsize=(16,9))
sns.set()
sns.displot(tips_df["total_bill"],rug_kws={'color':'#008000','edgecolor':'#FF0000','linewidth':5,'linestyle':'--','alpha':0.9}
           ,rug=True)
```

```python
plt.figure(figsize=(16,9))
sns.displot(tips_df['total_bill'],bins=9,label='total_bill')
sns.displot(tips_df['tip'],bins=9,label='tip')
sns.displot(tips_df['size'],bins=9,label='size')

plt.legend()
plt.show()
```
