# Zeros

```python
import numpy as np

ar_zero = np.zeros(4)
ar_zero1 = np.zeros((3,4))

print(ar_zero)
print()
print(ar_zero1)
```

```python
ar_zero1 = np.zeros((3,4),dtype =bool)
print(ar_zero1)
```

```python
ar_zero1 = np.zeros((3,4),dtype =str)
print(ar_zero1)
```

```python
ar_zero1 =''
print(bool(ar_zero1))
```

# ones

```python
ar_one = np.ones(4)
print(ar_one)
```

```python
ar_one.dtype
```

```python
ar_one = np.ones((3,4))
print(ar_one)
```

```python
ar_1 = np.ones((3,4),dtype = int)
print(ar_1)
```

# Empty

```python
ar_em = np.empty((3,3))

print(ar_em)
```

# Range

```python
ar_en = np.arange(1,13)
print(ar_en)
```

```python
even_ar =np.arange(1,13,2)
print(even_ar)
```

# Diagonal

```python
ar_dia = np.eye(3)
print(ar_dia)
```

```python
ar_dia = np.eye(3,5)
print(ar_dia)
```

# linspace

```python
ar_lin = np.linspace(0,20,num=5)

print(ar_lin)
```

# reshape()

```python
ar_2d = ar_en.reshape(3,4)
print(ar_2d)
```

```python
ar_3d = ar_en.reshape(2,3,2)
print(ar_3d)
```

```python
ar =np.arange(1,13).reshape(2,6)
print(ar)
```
