# Line Plot

```python
import matplotlib.pyplot as plt
```

```python
Days =[1,2,3,4,5,6,7,7,8,9]
Temp =[23,56,78,23,67,89,56,78,90,20]

plt.plot(Days,Temp)
plt.title("Delhi Temperature")
plt.xlabel("Days")
plt.ylabel("Temp")
plt.show()
```

```python
Days =[1,2,3,4,5,6,7,7,8,9]
Temp =[23,56,78,23,67,89,56,78,90,20]

plt.plot(Days,Temp,color ='r',marker ='o')
plt.axis([0,20,0,100])
plt.title("Delhi Temperature")
plt.xlabel("Days")
plt.ylabel("Temp")
plt.show()
```

```python
Days =[1,2,3,4,5,6,7,7,8,9]
Temp =[23,56,78,23,67,89,56,78,90,20]

plt.plot(Days,Temp,"go--",linewidth =2,markersize = 5)
plt.title("Delhi Temperature",fontsize =15)
plt.xlabel("Days",fontsize =10)
plt.ylabel("Temp",fontsize =10)
plt.legend(loc =4)
plt.show()
```

```python
from matplotlib import style
Days =[1,2,3,4,5,6,7,7,8,9]
Temp =[23,56,78,23,67,89,56,78,90,20]

style.use("ggplot")
plt.plot(Days,Temp,"ro--",linewidth =2,markersize = 5)
plt.title("Delhi Temperature",fontsize =15)
plt.xlabel("Days",fontsize =10)
plt.ylabel("Temp",fontsize =10)
plt.legend(loc =4)
plt.grid(color='k',linestyle = '-',linewidth=1)
plt.show()
```

```python
from matplotlib import style
Days =[1,2,3,4,5,6,7,7,8,9]
Delhi_Temp =[23,56,78,23,67,89,56,78,90,20]
Mumbai_Temp = [45,78,13,89,90,12,34,17,53,94]

style.use("ggplot")
plt.plot(Days,Delhi_Temp,"mo--",linewidth =2,markersize = 5,label = "Delhi_Temp")
plt.plot(Days,Mumbai_Temp,"ro--",linewidth =2,markersize = 5,label ="Mumbai_Temp")
plt.title("Delhi& Mumbai Temperature",fontsize =15)
plt.xlabel("Days",fontsize =10)
plt.ylabel("Temp",fontsize =10)
plt.legend(loc =4)
plt.grid(color='k',linestyle = '-',linewidth=1)
plt.show()
```
